```python
import numpy as np

# Creating dataset with 5 columns and 10 records (including some missing values)
data_cust = {
    'Customer_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Hannah', 'Ian', 'Jack'],
    'Age': [25, 34, np.nan, 28, 42, 31, 22, 39, 29, np.nan],
    'Gender': ['Female', 'Male', 'Male', 'Male', 'Female', 'Male', 'Female', 'Female', 'Male', 'Male'],
    'Purchase_Amount': [150.0, 230.5, 120.0, np.nan, 450.0, 310.2, 99.9, 520.0, 180.0, 210.0]
}

df_cust = pd.DataFrame(data_cust)

# 1. Display the first 5 rows using head()
print(" First 5 Rows")
print(df_cust.head())

# 2. Get summary statistics using describe()
print("\nSummary Statistics")
print(df_cust.describe())

# 3. Check for missing values
print("Missing Values Count ")
print(df_cust.isnull().sum())

# Handle missing values (filling numerical columns with median/mean)
df_cust['Age'] = df_cust['Age'].fillna(df_cust['Age'].median())
df_cust['Purchase_Amount'] = df_cust['Purchase_Amount'].fillna(df_cust['Purchase_Amount'].mean())
```

```python
# 1. Retrieve all customers whose Age is greater than 30
age_gt_30 = df_cust[df_cust['Age'] > 30]
print("Customers with Age > 30 ")
print(age_gt_30)

# 2. Find the total Purchase Amount for male and female customers separately
total_purchase_by_gender = df_cust.groupby('Gender')['Purchase_Amount'].sum()
print("\nTotal Purchase Amount by Gender")
print(total_purchase_by_gender)

# 3. Select only the Customer_ID, Name, and Purchase_Amount columns
selected_columns = df_cust[['Customer_ID', 'Name', 'Purchase_Amount']]
print("\nSelected Columns")
print(selected_columns)
```

```python
# 1. Add a new column "Discounted_Price" with a 10% discount
df_cust['Discounted_Price'] = df_cust['Purchase_Amount'] * 0.9

# 2. Group the data by Gender and calculate the average Purchase_Amount
avg_purchase_by_gender = df_cust.groupby('Gender')['Purchase_Amount'].mean()
print("Average Purchase Amount by Gender")
print(avg_purchase_by_gender)

# 3. Sort the dataset by Purchase_Amount in descending order
df_sorted = df_cust.sort_values(by='Purchase_Amount', ascending=False)
print("\n Sorted Dataset (Descending Purchase Amount)")
print(df_sorted)
```
