```python
filename="result.text"
words=["hello","world","i","love","programming"]
def zipper_file(filename,words):
    with open(filename, 'w') as f:
        for i in range(0, len(words), 2):
            zip_file=words[i]+words[i+1]

            f.write(zip_file + '\n')
            zipper_file(filename,words)


            
```

```python
dates=["01/15/2026","03/09/2026","05/21/2026","07/30/2026","09/12/2026"]
def reformat_date(dates):
    months=["january","february","march","april","may","june","july","august","september","october","november","december"]
    result=[]
    for date in dates:
        month,date,year=date.split("/")
        month_name =months[int(month)-1]

        sentence=f" Office Hours on {month_name}, {date},{year}"
        result.append(sentence)
        return result
    print(reformat_date(dates))
```
