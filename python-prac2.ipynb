```python
# 1. Create an empty list
shopping_list = []
print(shopping_list)

```

```python
# 2. Store the shopping list items
shopping_list = ["Watch", "laptop", "shoes", "pen", "clothes"]
print("Number of items:", len(shopping_list))
```

```python
# 3. Add new item "Football"
shopping_list.append("Football")
print(shopping_list)
```

```python

# 4. Print last item from the shopping list
print("Last item:", shopping_list[-1])
```

```python
# 5. Print entire shopping list
print("Shopping list:", shopping_list)
```

```python
# 6. Print important items (laptop and shoes)
important_items = [item for item in shopping_list if item in ("laptop", "shoes")]
print("Important items:", important_items)
```

```python
# 7. Change "pen" to "notebook"
pen_index = shopping_list.index("pen")
shopping_list[pen_index] = "notebook"
print("Updated list:", shopping_list)
```

```python
# 8. Delete the least important item and print the list
shopping_list.remove("clothes")  # assuming "clothes" is least important - change if it's a different item
print("Final shopping list:", shopping_list)
```
